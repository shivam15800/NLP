# Writing to a file

```python
# open('sample1.txt', 'w')
def demo():
    f1 = open('sample1.txt', 'w')
    f1.write("Hello NLP! \n")
    f1.write("This is file Handelling.")
    f1.close()
demo()
```

```python
def demo2():
    f2 = open('sample2.txt', 'w')
    for x in range(1, 10):
        f2.write(str(x))
        f2.write('\n')
    f2.close()
demo2()
```

```python
f3 = open('sample1.txt', 'r')
for text in f3:
    print(text)
```

# Reading Numbers From a File

```python
f = open('sample2.txt', 'r')
text = f.read()
print(text)
```

```python
f4 = open('sample2.txt', 'r')
total = 0
for i in range(10):
    line = f4.readline()
    if not line:   # EOF
        break
    total += int(line.strip())
print(total)
```
